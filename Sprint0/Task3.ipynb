### Вариант задания № 18 — Система отзывов


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создан базовый класс `Review`, представляющий отзыв о продукте или услуге. Класс содержит идентификатор отзыва `ReviewId`, текст отзыва `Text` и рейтинг `Rating`, а также методы `DisplayReview()`, `RateProduct()` и `GetReviewDetails()`.

На основе базового класса реализованы два производных класса: `ServiceReview` и `ProductReview`. `ServiceReview` содержит дополнительный атрибут `VisitDate` и переопределяет метод `DisplayReview()` для вывода даты посещения. `ProductReview` содержит атрибут `ProductId` и переопределяет метод `RateProduct()` для привязки выставленного рейтинга к конкретному продукту. Наследование и переопределение методов демонстрируют полиморфизм.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
Review[] reviews =
{
    new ServiceReview(1, "Хорошее обслуживание", 5, new DateTime(2026, 9, 9)),
    new ProductReview(2, "Качественный товар", 4, 101)
};

foreach (Review review in reviews)
{
    review.DisplayReview();
    Console.WriteLine(review.GetReviewDetails());
    Console.WriteLine();
}

ProductReview productReview = new ProductReview(3, "Неплохой товар", 3, 202);
productReview.RateProduct(5);
Console.WriteLine(productReview.GetReviewDetails());


// Базовый класс Review
class Review
{
    public int ReviewId { get; set; }
    public string Text { get; set; }
    public int Rating { get; protected set; }

    public Review(int reviewId, string text, int rating)
    {
        ReviewId = reviewId;
        Text = text;
        Rating = rating;
    }

    public virtual void DisplayReview()
    {
        Console.WriteLine($"Отзыв №{ReviewId}: {Text}. Рейтинг: {Rating}/5");
    }

    public virtual void RateProduct(int rating)
    {
        Rating = rating;
        Console.WriteLine($"Рейтингу присвоено значение: {Rating}/5");
    }

    public virtual string GetReviewDetails()
    {
        return $"ID отзыва: {ReviewId}; Текст: {Text}; Рейтинг: {Rating}/5";
    }
}


// Производный класс для отзыва об обслуживании
class ServiceReview : Review
{
    public DateTime VisitDate { get; set; }

    public ServiceReview(int reviewId, string text, int rating, DateTime visitDate)
        : base(reviewId, text, rating)
    {
        VisitDate = visitDate;
    }

    public override void DisplayReview()
    {
        Console.WriteLine(
            $"Отзыв об обслуживании №{ReviewId}: {Text}. " +
            $"Рейтинг: {Rating}/5. Дата посещения: {VisitDate:dd.MM.yyyy}");
    }

    public override string GetReviewDetails()
    {
        return base.GetReviewDetails() +
               $"; Дата посещения: {VisitDate:dd.MM.yyyy}";
    }
}


// Производный класс для отзыва о товаре
class ProductReview : Review
{
    public int ProductId { get; set; }

    public ProductReview(int reviewId, string text, int rating, int productId)
        : base(reviewId, text, rating)
    {
        ProductId = productId;
    }

    public override void RateProduct(int rating)
    {
        Rating = rating;
        Console.WriteLine(
            $"Товару с ID {ProductId} присвоен рейтинг: {Rating}/5");
    }

    public override string GetReviewDetails()
    {
        return base.GetReviewDetails() +
               $"; ID продукта: {ProductId}";
    }
}
```
